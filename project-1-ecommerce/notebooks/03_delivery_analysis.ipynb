```python
customers = pd.read_csv("/Users/aksharadarapaneni/data-analyst-portfolio/project-1-ecommerce/data/olist_customers_dataset.csv")
orders = pd.read_csv("/Users/aksharadarapaneni/data-analyst-portfolio/project-1-ecommerce/data/olist_orders_dataset.csv")
order_reviews= pd.read_csv("/Users/aksharadarapaneni/data-analyst-portfolio/project-1-ecommerce/data/olist_order_reviews_dataset.csv")
sellers= pd.read_csv("/Users/aksharadarapaneni/data-analyst-portfolio/project-1-ecommerce/data/olist_sellers_dataset.csv")
order_items = pd.read_csv("/Users/aksharadarapaneni/data-analyst-portfolio/project-1-ecommerce/data/olist_order_items_dataset.csv")
```

```python
orders['order_purchase_timestamp']=pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_approved_at']=pd.to_datetime(orders['order_approved_at'])
orders['order_delivered_carrier_date']=pd.to_datetime(orders['order_delivered_carrier_date'])
orders['order_delivered_customer_date']=pd.to_datetime(orders['order_delivered_customer_date'])
orders['order_estimated_delivery_date']=pd.to_datetime(orders['order_estimated_delivery_date'])
```

```python
orders_delivered=orders[orders['order_status']=='delivered'].copy()
orders_delivered.head()
```

```python
orders_delivered['delayed_delivered']=(orders_delivered['order_delivered_customer_date']-orders_delivered['order_estimated_delivery_date'])
orders_delivered.head()
```

```python
orders_delivered['delayed_days']=orders_delivered['delayed_delivered'].dt.days
orders_delivered['delayed_days'].head()
```

```python
late_delivered_orders=orders_delivered[orders_delivered['delayed_days']>=0].copy()
len(late_delivered_orders)
```

```python
late_pct = (7826 / len(orders_delivered)) * 100
print(f"Late orders: 7,826 ({late_pct:.1f}% of all delivered orders)")
```

```python
orders_with_reviews=order_reviews.merge(orders_delivered,on='order_id')
orders_with_reviews.head()
```

```python
late_review_score=orders_with_reviews[orders_with_reviews['delayed_days']>0]['review_score'].mean()
print(f"Late orders avg review: {late_review_score:.2f}")
```

```python
ontime_review_score=orders_with_reviews[orders_with_reviews['delayed_days']<=0]['review_score'].mean()
print(f"On-time orders avg review: {ontime_review_score:.2f}")
```

```python
orders_with_sellers= orders_delivered.merge(order_items,on='order_id')
```

```python
orders_with_sellers=orders_with_sellers.merge(sellers,on='seller_id')
```

```python

orders_with_sellers.columns
```

```python
orders_with_sellers.shape
```

```python
orders_with_sellers['is_late']=np.where(orders_with_sellers['delayed_days']>0,1,0)
```

```python
orders_with_sellers['is_late'].head()
```

```python
late_by_state=orders_with_sellers.groupby('seller_state')['is_late'].mean()*100
late_by_state=late_by_state.sort_values(ascending=False).head(10)
```

```python
plt.bar(late_by_state.index,late_by_state.values)
plt.xlabel('state')
plt.ylabel('late delivery %')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
